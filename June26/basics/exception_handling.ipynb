```python
# who defines functions or classes document exceptions
def divide(a:int|float, b: int|float) -> int|float:
    """This method divides two numbers

    Args:
        a (int|float): number
        b (int|float): number

    Raises:
       ZeroDivisionError when b == 0

    Returns:
        int|float: _description_
    """
    return a / b
```

```python
# if you are calling a functionality which might raise exceptions
# use try block if you want to handle it
try:
    divide(5,0)
except ZeroDivisionError as e:
    print(e)
    print("Cannot divide by zero")
```

```python
def simple_intrest(p,t,r):
    """This method calculates simple intrest

    Args:
        p (_type_): _description_
        t (_type_): _description_
        r (_type_): _description_

    Returns:
        _type_: _description_

    Raises:
        ValueError: if p, t, r are negative
    """
    if p < 0 or t < 0 or r < 0:
        # Raising an error
        raise ValueError("Principal, time and rate cannot be negative")
    return p * t * r / 100
```

```python
try:
    simple_intrest(-100, 2, 5)
    print("Inside try")
except ValueError as ve:
    print(ve)
    print("Inside except block")
finally:
    print("Finally block executed")
```

```python
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def transfer(self, other, amount):
        self.balance -= amount
        other.balance += amount

    def __str__(self):
        return f"{self.name} has {self.balance} balance"
```

```python
acc_1 = Account("akhil", 1000)
acc_2 = Account("bhavani", 2000)
```

```python
print(f"before transfer {acc_1}")
print(f"before transfer {acc_2}")
acc_2.transfer(acc_1, 500)
print(f"after transfer {acc_1}")
print(f"after transfer {acc_2}")
```

```python
class InsufficientBalanceError(Exception):
    pass

class ZeroBalanceError(Exception):
    pass

class KYCNotFoundError(Exception):
    pass
```

```python
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def transfer(self, other, amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient Balance, Kindly Deposit more funds")
        if self.balance - amount <= 0:
            raise ZeroBalanceError("Zero Balance: Sweeping account is not allowed")
        self.balance -= amount
        other.balance += amount

    def __str__(self):
        return f"{self.name} has {self.balance} balance"
```

```python
acc_1 = Account("akhil", 1000)
acc_2 = Account("bhavani", 2000)
```

```python
try:
    print(f"before transfer {acc_1}")
    print(f"before transfer {acc_2}")
    acc_1.transfer(acc_2, 20000)
    print(f"after transfer {acc_1}")
    print(f"after transfer {acc_2}")
except InsufficientBalanceError as ie:
    print(ie)
except ZeroBalanceError as ze:
    print(ze)
```

```python
try:
    print(f"before transfer {acc_1}")
    print(f"before transfer {acc_2}")
    acc_1.transfer(acc_2, 1000)
    print(f"after transfer {acc_1}")
    print(f"after transfer {acc_2}")
except InsufficientBalanceError as ie:
    print(ie)
except ZeroBalanceError as ze:
    print(ze)
```
