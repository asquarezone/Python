```python
class Student:
    school = "QT"

    def __init__(self, name, age, courses=[]):
        self.name = name
        self.age = age
        self.courses = courses

    # instance method
    def print_details(self):
        print(f"Name is {self.name} and age is {self.age}")
```

```python
s1 = Student("name1", 21)
s2 = Student("name2", 22)

# calling an instance method
s1.print_details()
# the above syntax is what we use the most

# other way of calling instance method
Student.print_details(s1)
```

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def print_details(self):
        print(f"Make is {self.make} and model is {self.model}")
```

```python
# Car is derived from Vehicle
# Vehicle => Parent
# Car => Child
class Car(Vehicle):
    pass
```

```python
c1 = Car(make="Hyundai", model="i20")
```

```python
c1.print_details()
```

```python
class Truck(Vehicle):
    
    def print_details(self):
        print(f"Truck Make is {self.make} and model is {self.model}")
```

```python
t1 = Truck(make="Tata", model="Sumo")
t1.print_details()
```

## Overriding behavior

* Lets consider different types of notifications

```python
class Notification:
    def send(self, message, sender, reciever):
        pass
```

```python
class EmailNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"Email sent from {sender} to {reciever} with message {message}")
        
class SMSNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"SMS sent from {sender} to {reciever} with message {message}")

class WhatsappNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"Whatsapp sent from {sender} to {reciever} with message {message}")

class PushNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"Push Notification sent from {sender} to {reciever} with message {message}")
```

```python
user1_notification_setting = [
    EmailNotification(),
    WhatsappNotification(),
    PushNotification(),
    WhatsappNotification()
]

for notification in user1_notification_setting:
    notification.send("Hello", "System", "abc@gmail.com")
```

```python
# super 

class Animal:
    def __init__(self, name):
        self.name = name
        print("Animal constructor")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name) # calling parent implementation from child
        self.breed = breed
        print("Dog constructor")
```

```python
dog = Dog(name="Tommy", breed="Husky")
print(dog.name)
print(dog.breed)
```

```python
class User:
    def __init__(self, username):
        self.username = username

    def login(self):
        print(f"{self.username} logged in")
```

```python
class AdminUser(User):
    def __init__(self, username, permissions):
        super().__init__(username)
        self.permissions = permissions

    def login(self):
        super().login()
        print(f"Admin permissions loaded: {self.permissions}")
```

```python
admin = AdminUser("admin", ["create_user", "delete_user", "modify_user", "get_user"])
```

```python
admin.login()
```

```python
user = User(username="khaja")
user.login()
```

```python
## Multi Level Inhertience

class AppUser:
    def login(self):
        print(f"User logged in")
    def logout(self):
        print("User logged out")

class StaffUser(AppUser):

    def mark_attendance(self):
        print("Staff marked attendance")


class AppAdminUser(StaffUser):
    def manage_users(self):
        print("Admin can manage user")
```

```python
# Multiple Inhertience


class Camera:
    def take_picture(self):
        print("Picture clicked")


class Phone:
    def make_call(self):
        print("making call")


class SmartPhone(Camera, Phone):
    def use_app(self):
        print("using app")
```

```python
s = SmartPhone()
s.make_call()
s.take_picture()
s.use_app()
```

```python
class Father:
    def speak(self):
        print("Father speaks")

    def play_football(self):
        print("Father plays football")

class Mother:
    def speak(self):
        print("Mother speaks")

    def sing(self):
        print("Mother sings")

class Child(Father, Mother):
    pass
```
