## Overriding behavior

* Lets consider different types of notifications

```python
class Notification:
    def send(self, message, sender, reciever):
        pass
```

```python
class EmailNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"Email sent from {sender} to {reciever} with message {message}")
        
class SMSNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"SMS sent from {sender} to {reciever} with message {message}")

class WhatsappNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"Whatsapp sent from {sender} to {reciever} with message {message}")

class PushNotification(Notification):
    def send(self, message, sender, reciever):
        print(f"Push Notification sent from {sender} to {reciever} with message {message}")
```

```python
user1_notification_setting = [
    EmailNotification(),
    WhatsappNotification(),
    PushNotification(),
    WhatsappNotification()
]

for notification in user1_notification_setting:
    notification.send("Hello", "System", "abc@gmail.com")
```

```python
# super 

class Animal:
    def __init__(self, name):
        self.name = name
        print("Animal constructor")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name) # calling parent implementation from child
        self.breed = breed
        print("Dog constructor")
```

```python
dog = Dog(name="Tommy", breed="Husky")
print(dog.name)
print(dog.breed)
```

```python
class User:
    def __init__(self, username):
        self.username = username

    def login(self):
        print(f"{self.username} logged in")
```

```python
class AdminUser(User):
    def __init__(self, username, permissions):
        super().__init__(username)
        self.permissions = permissions

    def login(self):
        super().login()
        print(f"Admin permissions loaded: {self.permissions}")
```

```python
admin = AdminUser("admin", ["create_user", "delete_user", "modify_user", "get_user"])
```

```python
admin.login()
```

```python
user = User(username="khaja")
user.login()
```

```python
## Dependency injection
from abc import ABC, abstractmethod
class NotificationSender(ABC):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def send(self, message, sender, reciever):
        pass

class EmailNotificationSender(NotificationSender):
    def send(self, message, sender, reciever):
        print(f"Email sent from {sender} to {reciever} with message {message}")

class SMSNotificationSender(NotificationSender):
    def send(self, message, sender, reciever):
        print(f"SMS sent from {sender} to {reciever} with message {message}")

class WhatsAppNotificationSender(NotificationSender):
    def send(self, message, sender, reciever):
        print(f"Whatsapp sent from {sender} to {reciever} with message {message}")

class UserNotitfication:
    def __init__(self, notification_sender):
        self.sender = notification_sender

    def notify(self, message, sender, reciever):
        self.sender.send(message, sender, reciever)
        
```

```python
my_notification = UserNotitfication(EmailNotificationSender(name="xyz"))
your_notification = UserNotitfication(SMSNotificationSender(name="abc"))
```

```python
my_notification.notify("User Logged in", "System", "abc@gmail.com")
```

```python
your_notification.notify("User Logged in", "System", "99999999")
```

```python
class DiscordNotificationSender(NotificationSender):
    def send(self, message, sender, reciever):
        print(f"Discord sent from {sender} to {reciever} with message {message}")
```

```python
xyz_notification = UserNotitfication(DiscordNotificationSender(name="xyz")) 
xyz_notification.notify("User Logged in", "System", "abc@gmail.com")
```

### Access modifiers

```python
# Everything is public
class MyClass:
    def __init__(self):
        self.name = "Public"

obj = MyClass()
obj.name
```

```python
# Everything is public
class MyClass:
    def __init__(self):
        self.name = "Public"
        # developer has used _ before so don't call from object
        self._age = 25

obj = MyClass()
print(obj.name)
print(obj._age) # It works, but you should not do it
```

```python
# Everything is public
class MyClass:
    def __init__(self):
        self.name = "Public"
        # developer has used _ before so don't call from object
        self._age = 25
        # this means private
        self.__balance = 100000

obj = MyClass()
print(obj.name)
print(obj._age) # It works, but you should not do it
obj.__balance
```

## Properties

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old"

p = Person(name="abc", age=21)
print(p)
p.age = -1
print(p)
```

```python
# age and name should not be changed after object creation

class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_age(self):
        return self.__age
    
    def set_age(self, age):
        if age <= 0:
            raise ValueError("Age should be greater than 0")
        self.__age = age

    def __str__(self):
        return f"{self.__name} is {self.__age} years old"

p = Person(name="abc", age=21)
p.set_age(-1)
p.get_age()
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # getter property
    @property
    def age(self):
        return self.__age
    
    @age.setter
    def age(self, age):
        if age <= 0:
            raise ValueError("Age should be greater than 0")
        self.__age = age
    
    @property
    def name(self):
        return self.__name
    
    @name.setter
    def name(self, name):
        self.__name = name

    def __str__(self):
        return f"{self.name} is {self.age} years old"

p = Person(name="abc", age=21)
print(p.age)
print(p.name)
p.age = 22
p.name = "abcd"
print(p.age)
print(p.name)
```

```python
class Point:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def __add__(self, other):
        return Point(self._x + other._x, self._y + other._y)

    def __str__(self):
        return f"Point({self._x}, {self._y})"
    
    @property
    def x(self):
        return self._x
    
    @x.setter
    def x(self, value):
        self._x = value

    @property
    def y(self):
        return self._y
    
    @y.setter
    def y(self, value):
        self._y = value
```
