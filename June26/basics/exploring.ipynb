```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    
    def __eq__(self, value):
        return self.x == value.x and self.y == value.y
    
    def __str__(self):
        return f"Point({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Point({self.x}, {self.y})"
```

```python
p1 = Point(3, 5)
p2 = Point(4,6)
p3 = Point(3, 5)
```

```python
p1 == p2
```

```python
p1 == p3
```

```python
p1 + p2
```

```python
class Point:

    @staticmethod
    def sine():
        return 0

    # class method
    @classmethod
    def cosine(cls):
        return 0
    
    # instance method
    def test(self):
        return 0

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __eq__(self, value):
        return self.x == value.x and self.y == value.y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    def __hash__(self):
        return hash((self.x, self.y))
```

```python
# to access class method object is not required
Point.cosine()
```

```python
# to access instance method object needs to be created
p = Point(2,5)
p.test()
```
