```python
class Animal:
    def eat(self):
        print("Animal is eating")

    def run(self):
        print("Animal is running")

    def make_sound(self):
        print("Animal is making sound")
```

```python
# created object 
a = Animal()

# executing capabilities
a.eat()
a.run()
a.make_sound()
```

```python
# Dog is an Animal 
# Dog is inherited from animal
class Dog(Animal):
    def make_sound(self):
        print("Dog is barking")
```

```python
class Cat:
    pass
```

```python
d = Dog()
d.eat()
d.run()
d.make_sound()
```

```python
c = Cat()
c.eat()
c.run()
```

```python
class Doctor():
    def walk(self):
        print("walking")
```

```python
class Monkey(Doctor):
    pass
```

```python
d = Doctor()
d.walk()
```

```python
m = Monkey()
m.walk()
```

```python
class Engine():
    def start(self):
        print("Engine has started")

    def stop(self):
        print("Engine has stopped")
```

```python
class Car():
    def __init__(self):
        # car has an engine
        self.engine = Engine()

    def start(self):
        print("........")
        self.engine.start()
        print("......")

```

```python
c = Car()
c.start()
```

```python
class Student:
    # class members
    institute_name = "QT"
    institute_address = "Hyd"

    # initializer
    def __init__(self, name, address):
        # self is the object
        self.name = name
        self.address = address
```

```python
s1 = Student(name="s1", address="abc")
s2 = Student(name="s2", address="abd")
```

```python
s1.name
```

```python
s1.address
```

```python
s2.address
```

```python
s2.name
```
