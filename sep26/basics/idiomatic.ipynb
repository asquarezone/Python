## Traditional approach

```python
squares = []

for i in range(1,6):
    squares.append(i * i)

squares
```

## Pythonic Approach

```python
squares = [ n * n for n in range(1,6) ]
squares
```

## Twelve idiomatic python techniques

### Technique 1 - Iterate over values not indexes

```python
fruits = ["apple", "mango", "kiwi"]

# not a good idea
for i in range(len(fruits)):
    print(fruits[i])

for fruit in fruits:
    print(fruit)
```

```python
## Technique 2 - using zip to combine collections

students = ["A", "B", "C"]
scores = [89, 76, 95]

dict(zip(students, scores))
```

```python
for name, score in zip(students, scores):
    print(f"{name} => {score}")
```

```python
# T-3 unpack values elegantly

employee = ("Andy", "Developer", 850000)
name, role, salary = employee
print(name)
print(role)
print(salary)
```

```python
# T-4

tasks = []

if len(tasks) == 0:
    print("Tasks unavailable")

if tasks:
    print("Tasks available")
else:
    print("Tasks unavailable")
```

```python
# T-5 use any or all

scores = [82, 74, 91, 33]

did_he_fail = any( score < 35 for score in scores)
print(did_he_fail)
```

```python
# T-5 use any or all

scores = [82, 74, 91, 83]

did_he_fail = any( score < 35 for score in scores)
print(did_he_fail)
```

```python
# T-5 use any or all

scores = [82, 74, 91, 83]

did_he_pass = all( score >= 35 for score in scores)
print(did_he_pass)
```

```python
## Building dictionaries using comprehensions

products = ["laptop", "mouse", "keyboard"]
prices = [65000, 300, 500]

catalog = {
    name: price
    for name, price in zip(products, prices)
}

catalog
```

```python
# T-7 count itmes with Counter

from collections import Counter

failures = ["timeout", "network", "timeout", "authentication", "network", "timeout"]

counts= Counter(failures)
counts
```

```python
counts.most_common(1)
```

```python
# T -8

from collections import defaultdict

records = [
    ("A", "Python"),
    ("B", "Java"),
    ("C", "FastAPI")
]

courses = defaultdict(list)

for student, course in records:
    courses[student].append(course)

courses
```

```python
courses["A"]
```
