```python
# let me add doc strings
def say_hello(name:str="python") -> None:
    """This function is used to say hello to a name

    Args:
        name (str, optional): name. Defaults to "python".

    Example
    """
    print(f"hello {name}")
```

```python
def is_even(number:int) -> bool:
    """This function checks if the number is even or not

    Args:
        number (int): number

    Returns:
        bool: True if even, False otherwise
    """
    return number % 2 == 0
```

```python
is_even(10)
```

```python
is_even(7)
```

```python
is_even("Hello")
```

```python
def add(a: int|float,b:int|float) -> int|float:
    """Adds two numbers

    Args:
        a (int | float): number 
        b (int | float): number

    Returns:
        int|float: sum of two numbers
    """
    return a + b
```

```python
add(3,5)
```

```python
add(3.5, 2.5)
```

```python
add("Hello", "World")
```
