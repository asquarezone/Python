## Generator

```python
# list all even numbers between some range

def list_even_numbers(start:int = 1, end:int=10):
    results = []
    for number in range(start,end+1):
        if number % 2 == 0:
            results.append(number)
    return results
```

```python
list_even_numbers()
```

```python
def even_numbers_generator(start:int = 1, end:int=10):
    for number in range(start, end+1):
        if number % 2 == 0:
            yield number
```

```python
for even_number in even_numbers_generator(1,10):
    print(even_number)
```
